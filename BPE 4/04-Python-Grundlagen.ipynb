# 2. Zeichenketten (Strings) und Variablen

Eine **Variable** ist ein Name, unter dem ein Wert gespeichert wird. Zeichenketten (`str`) enthalten Text und stehen in Anführungszeichen.

```python

```

Zeichenketten lassen sich mit `+` **konkatenieren**, also aneinanderhängen:

```python

print(vorname + " " + nachname)
```

Wichtig: `+` hat abhängig vom Datentyp eine unterschiedliche Bedeutung. Bei Strings verbindet es Texte; bei Zahlen addiert es.

````python
fach = "Informations"
profil = "technik"
bezeichnung = fach + profil
print(bezeichnung)
````

# 3. Ganzzahlen und Rechenoperatoren

Ganze Zahlen besitzen in Python den Datentyp `int`. Für Rechnungen stehen unter anderem folgende Operatoren zur Verfügung:

| Operator | Bedeutung |
|---|---|
| `+` | Addition |
| `-` | Subtraktion |
| `*` | Multiplikation |
| `/` | Division |
| `**` | Potenz |

Python beachtet die übliche Operatorrangfolge. Mit Klammern lässt sich die Reihenfolge ausdrücklich festlegen.

````python
x = 12
y = 5

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x ** 2)
print((x + y) * 2)
````

# 4. Datentypen und Typumwandlung

Wichtige Datentypen aus dem Kurs:

- `int` – ganze Zahlen
- `float` – Kommazahlen
- `str` – Zeichenketten
- `bool` – Wahrheitswerte (`True` oder `False`)

Mit `type()` kann der Datentyp eines Wertes untersucht werden.

Datentypen lassen sich häufig explizit konvertieren:


Die Zeichenkette muss zur gewünschten Umwandlung passen. `int("Hallo")` führt beispielsweise zu einem Fehler.

# 5. Sprechende Variablennamen

Variablennamen sollten ihren Inhalt möglichst klar beschreiben.

```python

# besser

```

Im Python-Stil werden Variablennamen üblicherweise klein und mit Unterstrichen geschrieben (**snake_case**).

Beispiele: `spieler_name`, `anzahl_tore`, `temperatur_celsius`.

Namen dürfen Buchstaben, Ziffern und `_` enthalten, dürfen aber nicht mit einer Ziffer beginnen. Python-Schlüsselwörter wie `if` sind keine geeigneten Variablennamen.

# 7. Boolesche Werte und Vergleiche

Der Datentyp `bool` besitzt genau zwei Werte:

```python
True
False
```

Vergleiche liefern Wahrheitswerte:

| Operator | Bedeutung |
|---|---|
| `>` | größer als |
| `<` | kleiner als |
| `>=` | größer oder gleich |
| `<=` | kleiner oder gleich |
| `==` | gleich |
| `!=` | ungleich |

Achtung: `=` ist eine **Zuweisung**, `==` ein **Vergleich**.

````python
alter = 17
print(alter >= 18)
print(alter == 17)
print(alter != 16)
````

# 8. Logische Operatoren: `not`, `and`, `or`

Boolesche Ausdrücke können kombiniert werden.

- `not A` kehrt den Wahrheitswert um.
- `A and B` ist nur wahr, wenn **beide** Ausdrücke wahr sind.
- `A or B` ist wahr, wenn **mindestens einer** der Ausdrücke wahr ist.

Typische Rangfolge: **Klammern → `not` → `and` → `or`**. Bei komplexeren Ausdrücken erhöhen Klammern die Lesbarkeit.

````python
alter = 18
hat_ausweis = True

darf_hinein = alter >= 18 and hat_ausweis
print(darf_hinein)

print(not darf_hinein)
````

# 9. Verzweigungen mit `if`

Mit einer Verzweigung wird Code nur dann ausgeführt, wenn eine Bedingung wahr ist.


Nach der Bedingung steht ein **Doppelpunkt**. Der zugehörige Codeblock wird **eingerückt**. Die Einrückung gehört zur Python-Syntax.

````python
punktzahl = 75

if punktzahl >= 50:
    print("Bestanden")
````

# 10. `if` – `else`

# 11. Mehrere Fälle mit `elif`

Mit `elif` können mehrere Bedingungen nacheinander geprüft werden. Sobald eine Bedingung wahr ist, wird ihr Block ausgeführt; die nachfolgenden Zweige dieser Kette werden übersprungen.

```python
punkte = 72

if punkte >= 90:
    print("sehr gut")
elif punkte >= 75:
    print("gut")
elif punkte >= 50:
    print("bestanden")
else:
    print("nicht bestanden")
```

Die **Reihenfolge der Bedingungen** ist wichtig.

````python
x = 5
y = 2

if x == 7:
    print("x ist 7")
elif y == 7:
    print("y ist 7")
elif x > 5 and y == 2:
    print("x > 5 und y == 2")
else:
    print("Keine Bedingung trifft zu")
````
